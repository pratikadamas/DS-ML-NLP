```python
df.drop("Total Volume", axis=1, inplace=True)
```

```python
#apply log transformation to skewed features
df1=df.copy()
df
```

```python


cols = [
    "AveragePrice",
    "f1",
    "f2",
    "f3",
    "Total Bags",
    "Small Bags",
    "Large Bags",
    "XLarge Bags"
]

df1[cols] = np.log1p(df1[cols])
```

```python
df1
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(8,4))
sns.boxplot(x=df1["AveragePrice"])

plt.title("Box Plot of Average Price")
plt.show()
```

```python
Q1 = df1["AveragePrice"].quantile(0.25)
Q3 = df1["AveragePrice"].quantile(0.75)

IQR = Q3 - Q1

upper = Q3 + 1.5 * IQR

count_upper_outliers = (df1["AveragePrice"] > upper).sum()

print("Upper IQR bound:", upper)
print("Number of points above upper bound:", count_upper_outliers)
```

```python
city_mean = df1.groupby("region")["AveragePrice"].mean()
city_quartile = pd.qcut(
    city_mean,
    4,
    labels=["cheap","mid_low","mid_high","premium"]

)
df1["region"] = df1["region"].map(city_quartile)
```

```python
df1.drop("Date", axis=1, inplace=True)
```

```python
df1
```

```python
from sklearn.preprocessing import OneHotEncoder
import pandas as pd

encoder = OneHotEncoder(sparse_output=False, handle_unknown='ignore')

# transform
encoded_data = encoder.fit_transform(df1[['region']])

# convert to dataframe with column names
encoded_df = pd.DataFrame(
    encoded_data,
    columns=encoder.get_feature_names_out(['region']),
    index=df1.index
)

# concat with original dataframe
df1 = pd.concat([df1, encoded_df], axis=1)

df1.head()
```

```python
y=df1["AveragePrice"]
y
```

```python
x=df1.iloc[:, df1.columns != "AveragePrice"]
x
```

```python
df1
```

```python
#train test split
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.25, random_state=42)
```

```python
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()
x_train=scaler.fit_transform(x_train)
x_test = scaler.transform(x_test)
```

```python
x_train
```

```python
x_test
```

```python
from sklearn.linear_model import LinearRegression

model=LinearRegression()
model.fit(x_train,y_train)
```

```python
y_pred = model.predict(x_test)
```

```python
from sklearn.metrics import mean_squared_error, r2_score
```

```python
#multiple linear regression
```

```python
print("MSE:", mean_squared_error(y_test, y_pred))
print("R2:", r2_score(y_test, y_pred))
```

```python
#polynominal regression
```

```python

from sklearn.pipeline import Pipeline
from sklearn.preprocessing import PolynomialFeatures, StandardScaler


# polynomial regression pipeline
polymodel = Pipeline([
    ("poly", PolynomialFeatures(degree=2, include_bias=False)),
    ("scaler", StandardScaler()),
    ("reg", LinearRegression())
])

# train
polymodel.fit(x_train, y_train)

# predict
y_pred_poly =polymodel.predict(x_test)
print("R2:", r2_score(y_test, y_pred_poly))
```

```python
from sklearn.linear_model import Ridge,Lasso,ElasticNet
```

```python
ridge_model = Ridge(alpha=0.00023)
ridge_model.fit(x_train, y_train)

```

```python
y_pred_ridge = ridge_model.predict(x_test)
print("R2:", r2_score(y_test, y_pred_ridge))
```

```python
lasso_model = Lasso(alpha=0.0009867)
lasso_model.fit(x_train, y_train)
```

```python
y_pred_lasso = lasso_model.predict(x_test)
print("R2:", r2_score(y_test, y_pred_lasso))
```

```python
elastinet_model = ElasticNet(alpha=0.009876, l1_ratio=0.5)
elastinet_model.fit(x_train, y_train)
```

```python
y_pred_lasso = lasso_model.predict(x_test)
print("R2:", r2_score(y_test, y_pred_lasso))
```
