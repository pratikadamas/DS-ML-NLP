```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error

# Sample data
X = np.array([1, 2, 3, 4, 5]).reshape(-1, 1)  # Feature (independent variable)
y = np.array([1.5, 3.2, 4.5, 6.1, 8.2])      # Target (dependent variable)

# Create and train the model
model = LinearRegression()
model.fit(X, y)

# Get the best fit line equation
slope = model.coef_[0]
intercept = model.intercept_

print(f"Best Fit Line: y = {slope:.2f}x + {intercept:.2f}")

# Predict values
y_pred = model.predict(X)

# Calculate Mean Squared Error
mse = mean_squared_error(y, y_pred)
print(f"Mean Squared Error: {mse:.4f}")

# Plot the data and the best fit line
plt.scatter(X, y, color='blue', label='Actual')
plt.plot(X, y_pred, color='red', label='Best Fit Line')
plt.xlabel('X')
plt.ylabel('y')
plt.legend()
plt.title('Simple Linear Regression')
plt.grid(True)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
from matplotlib import cm
from matplotlib.animation import FuncAnimation
from IPython.display import HTML

# Our 4D function: f(x,y,z) = x² + y² + z² + xy
def func(X):
    x, y, z = X
    return x**2 + y**2 + z**2 + x*y

# Gradient of the function
def gradient(X):
    x, y, z = X
    df_dx = 2*x + y
    df_dy = 2*y + x
    df_dz = 2*z
    return np.array([df_dx, df_dy, df_dz])

# Gradient descent implementation
def gradient_descent(start_point, learning_rate, max_iterations, tolerance):
    path = [start_point]
    costs = [func(start_point)]
    
    current_point = start_point.copy()
    
    for i in range(max_iterations):
        grad = gradient(current_point)
        new_point = current_point - learning_rate * grad
        path.append(new_point)
        costs.append(func(new_point))
        
        # Check for convergence
        if np.linalg.norm(new_point - current_point) < tolerance:
            break
            
        current_point = new_point
    
    return np.array(path), np.array(costs)

# Parameters
start_point = np.array([1.5, 1.5, 1.5])
learning_rate = 0.1
max_iterations = 50
tolerance = 1e-6

# Run gradient descent
path, costs = gradient_descent(start_point, learning_rate, max_iterations, tolerance)
print(path,"\n\nCosts:\n", costs)

## Visualization 1: 3D Parameter Space with Color as Cost
fig = plt.figure(figsize=(12, 10))

# Create a 3D grid for the background surface
x = np.linspace(-2, 2, 20)
y = np.linspace(-2, 2, 20)
z = np.linspace(-2, 2, 20)
X, Y, Z = np.meshgrid(x, y, z)
F = func([X, Y, Z])

# For visualization, we'll take a slice at z=0
slice_idx = np.argmin(np.abs(z))  # Index closest to z=0
X_slice = X[:, :, slice_idx]
Y_slice = Y[:, :, slice_idx]
F_slice = F[:, :, slice_idx]

ax1 = fig.add_subplot(221, projection='3d')
surf = ax1.plot_surface(X_slice, Y_slice, F_slice, cmap=cm.coolwarm, alpha=0.6)
ax1.set_title('Cost Surface Slice at z≈0')
ax1.set_xlabel('x')
ax1.set_ylabel('y')
ax1.set_zlabel('f(x,y,z)')

# Plot the optimization path (projected to z=0 for visualization)
path_proj = path.copy()
path_proj[:, 2] = 0  # Project z to 0 for this view
ax1.plot(path_proj[:, 0], path_proj[:, 1], func(path.T), 'r.-', linewidth=2, markersize=8)

## Visualization 2: Parameter Trajectory in 3D Space
ax2 = fig.add_subplot(222, projection='3d')
ax2.plot(path[:, 0], path[:, 1], path[:, 2], 'b.-', linewidth=2, markersize=8)
ax2.set_title('Parameter Space Trajectory')
ax2.set_xlabel('x')
ax2.set_ylabel('y')
ax2.set_zlabel('z')

# Mark start and end points
ax2.scatter(*path[0], color='g', s=100, label='Start')
ax2.scatter(*path[-1], color='r', s=100, label='End')
ax2.legend()

## Visualization 3: Cost over iterations
ax3 = fig.add_subplot(223)
ax3.plot(costs, 'b.-')
ax3.set_title('Cost Function Value vs Iteration')
ax3.set_xlabel('Iteration')
ax3.set_ylabel('Cost')
ax3.grid(True)

## Visualization 4: Animated Trajectory (3D)
fig_anim = plt.figure(figsize=(8, 6))
ax_anim = fig_anim.add_subplot(111, projection='3d')

def update(frame):
    ax_anim.clear()
    # Plot full path up to current frame
    ax_anim.plot(path[:frame+1, 0], path[:frame+1, 1], path[:frame+1, 2], 'b.-')
    # Current position
    ax_anim.scatter(path[frame, 0], path[frame, 1], path[frame, 2], color='r', s=100)
    ax_anim.set_xlim([-0.5, 1.5])
    ax_anim.set_ylim([-0.5, 1.5])
    ax_anim.set_zlim([-0.5, 1.5])
    ax_anim.set_title(f'Iteration {frame}\nCost: {costs[frame]:.4f}')
    ax_anim.set_xlabel('x')
    ax_anim.set_ylabel('y')
    ax_anim.set_zlabel('z')

anim = FuncAnimation(fig_anim, update, frames=len(path), interval=300, repeat=False)

plt.tight_layout()
plt.show()

# To display the animation in Jupyter notebook:
# HTML(anim.to_html5_video())

plt.rcParams['animation.writer'] = 'pillow'  # Requires pillow installed
HTML(anim.to_jshtml())  # Creates JavaScript animation
```
