##### Stemming
To normalize words so that similar words are treated the same.

Stemming is the process of reducing words to their root form (stem) by chopping off prefixes or suffixes.
It doesn’t always produce a real word, but it keeps words with similar meaning grouped together.

```python
# use in comment of products is a positive or negative review
#   eating ,eat eaten [similar meaning]
```

```python
words=["eat","eating","eaten","writing","write","written","fairly","fair","fairness","good","better","best","happy","happiness","happily"]
```

```python
## Techniques of Stemming
# 1. Porter Stemming
from nltk.stem import PorterStemmer
ps = PorterStemmer()
```

```python
for word in words:
    print(f"{word} --> {ps.stem(word)}")
```

```python
ps.stem("Congratulations")  # Congratulat

# meaning changed to congratul
```

```python
# 2.Regexp Stemming 
# regular_expression based stemming
from nltk.stem import RegexpStemmer
rs = RegexpStemmer('ing$|s$|e$|able$', min=4)
# where ever last 4 char are matching with ing,s,e,able that will be removed

rs.stem("ingeating")  # eat
  # be
```

```python
rs.stem("fairness") 
```

#### 3.snowball Stemming

```python
# 3.snowball Stemming
from nltk.stem import SnowballStemmer
ss = SnowballStemmer(language='english')
# ss.stem("happiness")  # happi
for word in words:
    print(f"{word} --> {ss.stem(word)}")
```

```python
# example of Porter Stemming
ps.stem("fairly"),ps.stem("sportingly")

```

```python
# Snowball Stemmer
ss.stem("fairly"),ss.stem("sportingly")
#  good results
```
