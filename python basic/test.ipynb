```python
#paking tuple
t2=1,2,3,"pratik girii"
print(f"Packed tuple: {t2}")
# unpacking tuple
a,b,c,d= t2
print(f"Unpacked values:\n a={a}\n b={b}\n c={c}\n d={d}")
```

```python
#unpacking with *
t3 = (1, 2, 3, 4, 5)
start, *middle, end,last = t3
print(f"Unpacked with *:\n start={start}\n middle={middle}\n end={end} \n last={last}")
type(middle)
```
