```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA

# 1. Generate correlated dataset
np.random.seed(42)

mean = [0, 0]
cov = [[3, 2],
       [2, 2]]  # covariance matrix (correlation)

X = np.random.multivariate_normal(mean, cov, 200)

# 2. Apply PCA
pca = PCA(n_components=2)
pca.fit(X)

components = pca.components_
variance = pca.explained_variance_

# 3. Plot dataset
plt.scatter(X[:,0], X[:,1])

# 4. Plot principal component directions
origin = np.mean(X, axis=0)

for i in range(2):
    vector = components[i] * np.sqrt(variance[i]) * 3
    plt.arrow(origin[0], origin[1],
              vector[0], vector[1],
              head_width=0.2,
              length_includes_head=True)

plt.title("PCA Visualization")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.show()
```

```python
pca_1d = PCA(n_components=1)
X_reduced = pca_1d.fit_transform(X)
X_back = pca_1d.inverse_transform(X_reduced)

plt.scatter(X[:,0], X[:,1], label="Original")
plt.scatter(X_back[:,0], X_back[:,1], label="Projected")
plt.legend()
plt.title("PCA Projection")
plt.show()
```
